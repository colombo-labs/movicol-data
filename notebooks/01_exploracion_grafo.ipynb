# 01 — Exploración del Grafo de Movilidad

Notebook básico para explorar el grafo construido de la red de transporte de Bogotá.

```python
import networkx as nx
import pandas as pd
from pathlib import Path

GRAPHS_DIR = Path("../data/graphs")
```

```python
# Cargar grafo enriquecido
G = nx.read_graphml(GRAPHS_DIR / "grafo_movilidad_bogota_enriched.graphml")
print(f"Nodos: {G.number_of_nodes():,}")
print(f"Aristas: {G.number_of_edges():,}")
print(f"Componentes conexas: {nx.number_connected_components(G)}")
```

```python
# Top 10 nodos por betweenness centrality
nodes_data = []
for node, data in G.nodes(data=True):
    nodes_data.append({"node": node, **data})

df = pd.DataFrame(nodes_data)
if "betweenness" in df.columns:
    df["betweenness"] = df["betweenness"].astype(float)
    print("Top 10 nodos más centrales:")
    display(df.nlargest(10, "betweenness")[["node", "name", "betweenness", "degree"]])
```

```python
# Distribución de grado
degrees = [G.degree(n) for n in G.nodes()]
print(f"Grado promedio: {sum(degrees)/len(degrees):.2f}")
print(f"Grado máximo: {max(degrees)}")
print(f"Grado mínimo: {min(degrees)}")
```

```python
# Componente principal
largest_cc = max(nx.connected_components(G), key=len)
pct = len(largest_cc) / G.number_of_nodes() * 100
print(f"Componente principal: {len(largest_cc):,} nodos ({pct:.1f}%)")
```
